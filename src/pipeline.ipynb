```python
print("baris sebelum drop transaction_date", len(df))
df = df.dropna(subset=["transaction_date"])
print("baris setelah drop transaction_date", len(df))
```

```python
df["transaction_date"].isna().sum()
```

```python
df[["make", "num-of-doors"]] = df[["make", "num-of-doors"]].fillna("unknown")
```

```python
df["make"].isna().sum()
```

```python
df["num-of-doors"].isna().sum()
```

```python
pd.crosstab(
    pd.cut(df["horsepower"], bins=10),
    df["horsepower-binned"]
)
```

```python
df[["horsepower", "horsepower-binned"]] \
    .dropna() \
    .sort_values("horsepower") \
    .drop_duplicates() \
    .loc[lambda x: x["horsepower-binned"].ne(
        x["horsepower-binned"].shift()
    )]
```

```python
df.groupby("horsepower-binned")["horsepower"].agg(
    ["min", "max"]
)
```

```python
df["horsepower-binned"] = df["horsepower-binned"].fillna(
    pd.cut(
        df["horsepower"],
        bins=[-float("inf"), 101, 155, float("inf")],
        labels=["Low", "Medium", "High"]
    )
)
```

```python
df["horsepower-binned"].isna().sum()
```

```python
df["stroke"] = df["stroke"].fillna(df["stroke"].mean())
```

```python
df["stroke"].isna().sum()
```

```python
df["price"].skew()
```

```python
df["price"] = df["price"].fillna(df["price"].median())
```

```python
df["price"].isna().sum()
```

```python
df["horsepower"].skew()
```

```python
df["horsepower"] = df["horsepower"].fillna(
    df.groupby("horsepower-binned", observed=True)["horsepower"]
      .transform("median")
)
```

```python
df["horsepower"].isna().sum()
```

```python
df = df.drop_duplicates()
```

```python
len(df)
```

```python
df["transaction_date"] = pd.to_datetime(
    df["transaction_date"], errors="coerce", format="mixed"
)
```

```python
cols = ["make", "body-style", "drive-wheels", "fuel-system"]
df[cols] = df[cols].apply(lambda col: col.str.lower().str.strip())
```

```python
mapping = {
    "one": 1,
    "two": 2,
    "three": 3,
    "four": 4,
    "five": 5,
    "six": 6,
    "seven": 7,
    "eight": 8,
    "nine": 9,
    "ten": 10,
    "eleven": 11,
    "twelve": 12,
    "unknown": -1
}

cols = ["num-of-doors", "num-of-cylinders"]

df[cols] = df[cols].replace(mapping)
```
